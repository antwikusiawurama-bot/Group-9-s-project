```python
import pandas as pd
import numpy as np
```

```python
utilities = pd.read_csv("../../../Raw/utilities.csv")
substations = pd.read_csv("../../../Raw/substations.csv")
lines = pd.read_csv("../../../Raw/lines.csv")
```

```python
print("Utilities duplicates:", utilities.duplicated().sum())
print("Substations duplicates:", substations.duplicated().sum())
print("Lines duplicates:", lines.duplicated().sum())
```

```python
# Remove duplicates (if any)
utilities = utilities.drop_duplicates()
substations = substations.drop_duplicates()
lines = lines.drop_duplicates()
```

```python
utilities.to_csv("../../utilities_cleaned.csv", index=False)
substations.to_csv("../../substations_cleaned.csv", index=False)
lines.to_csv("../../lines_cleaned.csv", index=False)

print("Cleaned datasets saved successfully!")
```

```python
# Verify Source and Destination Substation IDs exist

valid_ids = set(substations["Substation ID"])

invalid_source = lines[~lines["Source Substation ID"].isin(valid_ids)]
invalid_destination = lines[~lines["Destination Substation ID"].isin(valid_ids)]

print("Invalid Source IDs:", len(invalid_source))
print("Invalid Destination IDs:", len(invalid_destination))
```

```python
# Validate coordinates

invalid_coordinates = substations[
    (substations["Latitude"] < -5) |
    (substations["Latitude"] > 20) |
    (substations["Longitude"] < -20) |
    (substations["Longitude"] > 10)
]

print("Invalid coordinate records:", len(invalid_coordinates))
```
